## NORMALIZACIÓN DE 'academic_performance'

```python
# Normalizar la columna 'academic_performance' usando Min-Max scaling
min_val = df['academic_performance'].min()
max_val = df['academic_performance'].max()
df['academic_performance_normalized'] = (df['academic_performance'] - min_val) / (max_val - min_val)

df['academic_performance'] = df['academic_performance_normalized']
df = df.drop(columns=['academic_performance_normalized'])

display(df['academic_performance'].describe())
```
