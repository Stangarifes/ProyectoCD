```python
import pandas as pd

# Cargar el dataset
df = pd.read_csv('student_mental_health_burnout_1M.csv')

# Estadísticas descriptivas básicas
print("=== ESTADÍSTICAS DESCRIPTIVAS ===\n")
print(df.describe())

# Distribución de la variable objetivo
print("=== DISTRIBUCIÓN DE risk_level ===\n")
print(df['risk_level'].value_counts())
print("\n=== PROPORCIÓN (% ) ===\n")
print(df['risk_level'].value_counts(normalize=True) * 100)


```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar estilo de los gráficos (fondo limpio)
sns.set_style("whitegrid")

# 1. Gráfico de barras de la variable objetivo risk_level
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='risk_level', order=['Low', 'Medium', 'High'], palette='pastel')
plt.title('Distribución de Niveles de Riesgo de Burnout')
plt.xlabel('Nivel de riesgo')
plt.ylabel('Cantidad de estudiantes')
plt.tight_layout()
plt.savefig('distribucion_risk_level.png')  # Guarda la imagen
plt.show()

# 2. Histograma de la variable stress_level (nivel de estrés)
plt.figure(figsize=(6, 4))
plt.hist(df['stress_level'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribución del Nivel de Estrés')
plt.xlabel('Nivel de estrés (0-10)')
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.savefig('histograma_stress.png')
plt.show()

# 3. Histograma de sleep_hours (horas de sueño)
plt.figure(figsize=(6, 4))
plt.hist(df['sleep_hours'], bins=15, color='lightgreen', edgecolor='black')
plt.title('Distribución de Horas de Sueño por Noche')
plt.xlabel('Horas de sueño')
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.savefig('histograma_sleep.png')
plt.show()

# 4. Boxplot (diagrama de caja) para comparar estrés según nivel de riesgo
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x='risk_level', y='stress_level', order=['Low', 'Medium', 'High'], palette='Set2')
plt.title('Nivel de Estrés por Categoría de Riesgo')
plt.xlabel('Nivel de riesgo')
plt.ylabel('Nivel de estrés')
plt.tight_layout()
plt.savefig('boxplot_stress_por_riesgo.png')
plt.show()
```
